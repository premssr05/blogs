```python
import markdown, pathlib

template = """<!DOCTYPE html>
<html>
<head>
  <meta charset="utf-8">
  <meta name="viewport" content="width=device-width, initial-scale=1">
  <link rel="stylesheet" href="/style.css">
  <title>{title}</title>
</head>
<body>
{body}
</body>
</html>"""

path = "./blogs/personalblog"

for md_file in pathlib.Path(path).glob("*.md"):
    content = md_file.read_text()

    print(content)
    body = markdown.markdown(content)
    html = template.format(title=md_file.stem, body=body)
    pathlib.Path(f"save",".html").write_text(html)
```
