```python
import html
import markdown
from pathlib import Path

template = """<!DOCTYPE html>
<html>
<head>
  <meta charset="utf-8">
  <meta name="viewport" content="width=device-width, initial-scale=1">
  <title>{title}</title>
  <style>
    body {{ background-color: lightgreen; color: black; }}
    h1, h2, h3, h4, h5, h6 {{ background-color: blue; color: white; padding: 12px; }}
    .home-link {{ display: inline-block; margin: 12px; font-size: 1.5rem; }}
  </style>
</head>
<body>
  <a class="home-link" href="{home_href}" aria-label="Home" title="Home">&#x1F3E0;</a>
{body}
</body>
</html>"""

source_dir = Path("E:/blogs/personalblog")
output_dir = source_dir / "public"

for md_file in source_dir.rglob("*.md"):
    relative_md = md_file.relative_to(source_dir)
    relative_html = relative_md.with_suffix(".html")
    output_file = output_dir / relative_html
    output_file.parent.mkdir(parents=True, exist_ok=True)

    depth = len(relative_html.parts) - 1
    home_href = "../" * depth + "index.html"
    body = markdown.markdown(md_file.read_text(encoding="utf-8"))
    page = template.format(
        title=html.escape(md_file.stem), body=body, home_href=home_href
    )
    output_file.write_text(page, encoding="utf-8")
    print(f"Generated {output_file}")
```
